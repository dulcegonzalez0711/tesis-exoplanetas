### Criterios de inclusión y exclusión

Criterios de inclusión:
- Planetas presentes en la tabla `Planetary Systems Composite Parameters`.
- Planetas con nombre identificable (`pl_name`).
- Sistemas con estrella anfitriona identificable (`hostname`).
- Variables seleccionadas relevantes para el análisis poblacional.

Criterios de exclusión:
- Planetas marcados como controversiales (`pl_controv_flag = 1`).
- Filas duplicadas completas, si existieran.
- Registros con valores físicamente imposibles en variables principales, si existieran.


```python
def limpiar_catalogo_exoplanetas(ruta_raw):
    """
    Carga el catálogo original del NASA Exoplanet Archive,
    selecciona variables relevantes, revisa consistencia básica
    y excluye planetas controversiales.

    Parámetros
    ----------
    ruta_raw : str o Path
        Ruta al archivo CSV original.

    Retorna
    -------
    df_raw : DataFrame
        Catálogo original completo.
    df_work : DataFrame
        Catálogo con variables seleccionadas.
    df_clean : DataFrame
        Catálogo limpio base sin planetas controversiales.
    tabla_missing : DataFrame
        Tabla de valores faltantes por variable.
    tabla_muestras : DataFrame
        Tabla de tamaños de submuestras.
    tabla_checks_fisicos : DataFrame
        Tabla de revisiones físicas básicas.
    """

    # 1. Cargar catálogo original
    df_raw = pd.read_csv(ruta_raw, comment="#")

    # 2. Variables seleccionadas
    cols_trabajo = [
        # Identificación
        "pl_name", "hostname", "pl_letter",

        # Sistema
        "sy_snum", "sy_pnum", "sy_mnum", "cb_flag",

        # Descubrimiento
        "discoverymethod", "disc_year", "disc_facility",
        "rv_flag", "tran_flag", "micro_flag", "ima_flag", "ttv_flag",
        "pl_controv_flag",

        # Parámetros orbitales
        "pl_orbper", "pl_orbsmax", "pl_orbeccen",

        # Parámetros planetarios
        "pl_rade", "pl_bmasse", "pl_bmassprov", "pl_dens",
        "pl_insol", "pl_eqt",

        # Parámetros estelares
        "st_spectype", "st_teff", "st_rad", "st_mass", "st_met",
        "st_metratio", "st_lum", "st_logg", "st_age",

        # Distancia
        "sy_dist"
    ]

    df_work = df_raw[cols_trabajo].copy()

    # 3. Tabla de valores faltantes
    tabla_missing = pd.DataFrame({
        "variable": df_work.columns,
        "datos_disponibles": df_work.notna().sum().values,
        "datos_faltantes": df_work.isna().sum().values,
        "porcentaje_faltante": (df_work.isna().mean().values * 100).round(3)
    }).sort_values("porcentaje_faltante", ascending=False)

    # 4. Tabla de submuestras
    muestras = {
        "muestra_total": len(df_work),
        "sin_controversiales": len(df_work[df_work["pl_controv_flag"] == 0]),
        "con_radio": df_work["pl_rade"].notna().sum(),
        "con_masa": df_work["pl_bmasse"].notna().sum(),
        "con_densidad_reportada": df_work["pl_dens"].notna().sum(),
        "con_radio_y_masa": df_work[["pl_rade", "pl_bmasse"]].dropna().shape[0],
        "con_metalicidad": df_work["st_met"].notna().sum(),
        "con_masa_estelar": df_work["st_mass"].notna().sum(),
        "con_temperatura_estelar": df_work["st_teff"].notna().sum(),
        "con_radio_masa_metalicidad": df_work[["pl_rade", "pl_bmasse", "st_met"]].dropna().shape[0],
    }

    tabla_muestras = pd.DataFrame(
        muestras.items(),
        columns=["submuestra", "numero_de_planetas"]
    )

    # 5. Revisión de valores físicos imposibles
    checks_fisicos = {
        "radio_planetario_menor_o_igual_cero": (df_work["pl_rade"] <= 0).sum(),
        "masa_planetaria_menor_o_igual_cero": (df_work["pl_bmasse"] <= 0).sum(),
        "densidad_menor_o_igual_cero": (df_work["pl_dens"] <= 0).sum(),
        "periodo_orbital_menor_o_igual_cero": (df_work["pl_orbper"] <= 0).sum(),
        "semieje_mayor_menor_o_igual_cero": (df_work["pl_orbsmax"] <= 0).sum(),
        "temperatura_estelar_menor_o_igual_cero": (df_work["st_teff"] <= 0).sum(),
        "masa_estelar_menor_o_igual_cero": (df_work["st_mass"] <= 0).sum(),
        "radio_estelar_menor_o_igual_cero": (df_work["st_rad"] <= 0).sum(),
        "distancia_menor_o_igual_cero": (df_work["sy_dist"] <= 0).sum(),
    }

    tabla_checks_fisicos = pd.DataFrame(
        checks_fisicos.items(),
        columns=["revision", "numero_de_casos"]
    )

    # 6. Revisión de duplicados
    if df_work.duplicated().sum() > 0:
        df_work = df_work.drop_duplicates().copy()

    # 7. Excluir planetas controversiales
    df_clean = df_work[df_work["pl_controv_flag"] == 0].copy()

    return df_raw, df_work, df_clean, tabla_missing, tabla_muestras, tabla_checks_fisicos
```

```python
ruta_raw = Path("../data_raw/nasa_exoplanets_raw_2026_07_21.csv")

df_raw_test, df_work_test, df_clean_test, missing_test, muestras_test, checks_test = limpiar_catalogo_exoplanetas(ruta_raw)
```
